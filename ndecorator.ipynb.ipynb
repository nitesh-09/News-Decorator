```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
```

```python
data = pd.read_csv("fake_or_real_news.csv")
data
```

```python
data['R/F'] = data['label'].apply(lambda x: 0 if x=="REAL" else 1)
```

```python
x, y = data['text'], data ['R/F']
```

```python
x, y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y , test_size=0.2)
```

```python
vectorizer = TfidfVectorizer(stop_words="english", max_df=0.7)
x_train_vectorizer = vectorizer.fit_transform(x_train)
x_test_vectorizer= vectorizer.transform(x_test)
# we don't have to Vectorized Y as it is  int value
```

```python
clf = LinearSVC()
clf.fit(x_train_vectorizer, y_train)
```

```python
clf.score(x_train_vectorizer, y_train)
```

```python
clf.score(x_test_vectorizer, y_test)
```
